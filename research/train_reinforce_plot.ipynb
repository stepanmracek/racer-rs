```python
from matplotlib import pyplot as plt
import pandas as pd
```

```python
data = pd.read_csv("policy-reinforce/log.csv", names=("episode", "reward", "running_mean_reward"))
display(data)
```

```python
plt.plot(data.episode, data.reward, alpha=0.5)
plt.plot(data.episode, data.running_mean_reward)
plt.show()
```

```python
ix = data.episode % 100 == 0
plt.plot(data.episode[ix], data.reward[ix], alpha=0.5)
plt.plot(data.episode[ix], data.running_mean_reward[ix])
plt.show()
```
